```python
# use python weather API to add columns for weather data to the eurogate_all_snapshots.csv file
import requests
import json
import pandas as pd
import numpy as np

INPUT_CSV = "data/eurogate_train.csv"
OUTPUT_CSV = "data/eurogate_train.csv"

# Eurogate Hamburg / Port of Hamburg approximate coordinates
LAT = 53.541
LON = 9.958
TIMEZONE = "Europe/Berlin"

HOURLY_VARS = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "weather_code",
    "cloud_cover",
    "wind_speed_10m",
    "wind_gusts_10m"
]

df = pd.read_csv(INPUT_CSV)

df["weather_match_time"] = pd.to_datetime(df["arrivalTime"], errors="coerce", utc=True)

df["weather_hour"] = (
    df["weather_match_time"]
    .dt.tz_convert(TIMEZONE)
    .dt.tz_localize(None)
    .dt.round("h")
)

start_date = df["weather_hour"].min().date().isoformat()
end_date = df["weather_hour"].max().date().isoformat()

print(f"Fetching weather from {start_date} to {end_date}")

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": LAT,
    "longitude": LON,
    "start_date": start_date,
    "end_date": end_date,
    "hourly": ",".join(HOURLY_VARS),
    "timezone": TIMEZONE,
    "temperature_unit": "celsius",
    "wind_speed_unit": "kmh",
    "precipitation_unit": "mm",
}

response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

weather = pd.DataFrame(response.json()["hourly"])
weather["weather_time"] = pd.to_datetime(weather["time"])
weather = weather.drop(columns=["time"])

df_weather = df.merge(
    weather,
    left_on="weather_hour",
    right_on="weather_time",
    how="left"
)

df_weather["is_raining"] = (
    df_weather["precipitation"].fillna(0) > 0
).astype(int)

df_weather["bad_weather_score"] = (
    df_weather["is_raining"]
    + (df_weather["wind_gusts_10m"].fillna(0) > 30).astype(int)
    + (df_weather["cloud_cover"].fillna(0) > 80).astype(int)
)

df_weather.to_csv(OUTPUT_CSV, index=False)

print("Saved:", OUTPUT_CSV)

print("\nWeather missing fraction:")
weather_cols = HOURLY_VARS + ["is_raining", "bad_weather_score"]
print(df_weather[weather_cols].isna().mean().sort_values(ascending=False))

df_weather.head()
```

```python
import numpy as np
import pandas as pd

# typeCode: 0-10 back to original labels
type_reverse = {
    0: "TK",
    1: "RF",
    2: "DC",
    3: "OT",
    4: "FC",
    5: "DO",
    6: "SR",
    7: "PL",
    8: "AC",
    9: "OH",
    10: "FR",
}

# arrivalType: 0-3 back to labels
arrival_reverse = {
    0: "OCEAN_CARRIER",
    1: "FEEDER",
    2: "SCHUTE",
    3: "BINNEN",
}

# departureType: 1-7 back to labels
departure_reverse = {
    1: "TRUCK",
    2: "RAIL",
    3: "OCEAN_CARRIER",
    4: "INTER_TERMINAL",
    5: "FEEDER",
    6: "SCHUTE",
    7: "BINNEN",
}

# iedCode: 0/1 back to labels
ied_reverse = {
    0: "IMPORT",
    1: "EXPORT",
}

# released: if you encoded Y/N as 1/0
released_reverse = {
    1: "Y",
    0: "N",
}

# Apply safely
df["typeCode"] = df["typeCode"].map(type_reverse)
df["arrivalType"] = df["arrivalType"].map(arrival_reverse)
df["departureType"] = df["departureType"].map(departure_reverse)
df["iedCode"] = df["iedCode"].map(ied_reverse)
df["released"] = df["released"].map(released_reverse)

print(df[["typeCode", "arrivalType", "departureType", "iedCode", "released"]].head())
```

```python
# one hot encode the top twenty ISO codes (99.6% of data) and labels the rest as other
# this will add ~ 20 columns to dataset 
top20_iso = df["sizetypeIsoCode"].value_counts().nlargest(20).index

df["sizetypeIsoCode_clean"] = np.where(
    df["sizetypeIsoCode"].isin(top20_iso),
    df["sizetypeIsoCode"],
    "OTHER"
)

df = pd.get_dummies(
    df,
    columns=["sizetypeIsoCode_clean"],
    prefix="iso",
    drop_first=True
)

df = df.drop(columns=["sizetypeIsoCode"])
```

```python
# break arrival time into hour, day of week, month, weekend, and sin/cos for month/hour 
# -> so it is clear to the model that 12:00 -> 1:00 is 1 hour
df["arrivalTime"] = pd.to_datetime(df["arrivalTime"], errors="coerce", utc=True)

df["arrival_hour"] = df["arrivalTime"].dt.hour
df["arrival_dayofweek"] = df["arrivalTime"].dt.dayofweek
df["arrival_month"] = df["arrivalTime"].dt.month
df["arrival_is_weekend"] = (df["arrival_dayofweek"] >= 5).astype(int)

df["arrival_hour_sin"] = np.sin(2 * np.pi * df["arrival_hour"] / 24)
df["arrival_hour_cos"] = np.cos(2 * np.pi * df["arrival_hour"] / 24)

df["arrival_month_sin"] = np.sin(2 * np.pi * df["arrival_month"] / 12)
df["arrival_month_cos"] = np.cos(2 * np.pi * df["arrival_month"] / 12)
```

```python
# we only need eta to find "lateness" so make this column and then drop arrival ETA col

df["arrivalVoyageEta"] = pd.to_datetime(df["arrivalVoyageEta"], errors="coerce", utc=True)

df["eta_error_hours"] = (
    df["arrivalTime"] - df["arrivalVoyageEta"]
).dt.total_seconds() / 3600

df = df.drop(columns=["arrivalVoyageEta"])
```

```python
# dont need arrival time anymore 
df = df.drop(columns=["arrivalTime"])
```

```python
# too many categories so frequency encode instead for arrival pol code
df["arrivalPolCode"] = df["arrivalPolCode"].fillna("MISSING")

arrival_pol_freq = df["arrivalPolCode"].value_counts(normalize=True)

df["arrivalPolCode_freq"] = df["arrivalPolCode"].map(arrival_pol_freq)

df = df.drop(columns=["arrivalPolCode"])
```

```python
# also frequency encode for arrival vessel name
df["arrivalVesselName"] = df["arrivalVesselName"].fillna("MISSING")

vessel_freq = df["arrivalVesselName"].value_counts(normalize=True)

df["arrivalVesselName_freq"] = df["arrivalVesselName"].map(vessel_freq)

df = df.drop(columns=["arrivalVesselName"])
```

```python
# one hot encode the top 41 arrival service names 
# this will add 41 columns to dataset 

df["arrivalServiceName"] = df["arrivalServiceName"].fillna("MISSING")

df = pd.get_dummies(
    df,
    columns=["arrivalServiceName"],
    prefix="service",
    drop_first=True
)
```

```python
import pandas as pd
import numpy as np

RAW_CSV = "data/eurogate_all_snapshots.csv"
OUTPUT_CSV = "data/eurogate_predeparture_remaining_dwell.csv"

df_raw = pd.read_csv(RAW_CSV)

# -----------------------------
# Datetime setup
# -----------------------------
df_raw["arrivalTime"] = pd.to_datetime(df_raw["arrivalTime"], errors="coerce", utc=True)
df_raw["departureTime"] = pd.to_datetime(df_raw["departureTime"], errors="coerce", utc=True)
df_raw["arrivalVoyageEta"] = pd.to_datetime(df_raw["arrivalVoyageEta"], errors="coerce", utc=True)

df_raw["snapshot_time"] = (
    df_raw["source_file"]
    .str.extract(r"Containerdaten-(\d{4}-\d{2}-\d{2}_\d{2}\.\d{2}\.\d{2})")[0]
    .str.replace("_", " ", regex=False)
    .str.replace(".", ":", regex=False)
)

df_raw["snapshot_time"] = pd.to_datetime(
    df_raw["snapshot_time"],
    errors="coerce",
    utc=True
)

# -----------------------------
# Get final departure time per container
# -----------------------------
final_departure = (
    df_raw.dropna(subset=["departureTime"])
    .groupby("containerId")["departureTime"]
    .min()
    .rename("final_departureTime")
)

df_raw = df_raw.merge(
    final_departure,
    on="containerId",
    how="left"
)

# -----------------------------
# Keep only pre-departure snapshots
# -----------------------------
df_pre = df_raw[
    df_raw["departureTime"].isna()
    & df_raw["final_departureTime"].notna()
    & (df_raw["snapshot_time"] < df_raw["final_departureTime"])
].copy()

print("Pre-departure rows:", len(df_pre))
print("Unique containers:", df_pre["containerId"].nunique())

if len(df_pre) == 0:
    raise ValueError("No valid pre-departure snapshots found.")

# -----------------------------
# Target variables
# -----------------------------
df_pre["remaining_dwell_hours"] = (
    df_pre["final_departureTime"] - df_pre["snapshot_time"]
).dt.total_seconds() / 3600

df_pre["dwell_hours"] = (
    df_pre["final_departureTime"] - df_pre["arrivalTime"]
).dt.total_seconds() / 3600

# -----------------------------
# Arrival time features
# -----------------------------
df_pre["arrival_hour"] = df_pre["arrivalTime"].dt.hour
df_pre["arrival_dayofweek"] = df_pre["arrivalTime"].dt.dayofweek
df_pre["arrival_month"] = df_pre["arrivalTime"].dt.month
df_pre["arrival_is_weekend"] = (df_pre["arrival_dayofweek"] >= 5).astype(int)

df_pre["arrival_hour_sin"] = np.sin(2 * np.pi * df_pre["arrival_hour"] / 24)
df_pre["arrival_hour_cos"] = np.cos(2 * np.pi * df_pre["arrival_hour"] / 24)
df_pre["arrival_month_sin"] = np.sin(2 * np.pi * df_pre["arrival_month"] / 12)
df_pre["arrival_month_cos"] = np.cos(2 * np.pi * df_pre["arrival_month"] / 12)

df_pre["eta_error_hours"] = (
    df_pre["arrivalTime"] - df_pre["arrivalVoyageEta"]
).dt.total_seconds() / 3600

# -----------------------------
# Snapshot time features
# -----------------------------
df_pre["arrival_year"] = df_pre["arrivalTime"].dt.year
df_pre["snapshot_year"] = df_pre["snapshot_time"].dt.year
df_pre["snapshot_date"] = df_pre["snapshot_time"].dt.strftime("%Y-%m-%d")
df_pre["snapshot_hour"] = df_pre["snapshot_time"].dt.hour
df_pre["snapshot_minute"] = df_pre["snapshot_time"].dt.minute
df_pre["snapshot_second"] = df_pre["snapshot_time"].dt.second

df_pre["elapsed_dwell_hours"] = (
    df_pre["snapshot_time"] - df_pre["arrivalTime"]
).dt.total_seconds() / 3600

df_pre = df_pre[df_pre["elapsed_dwell_hours"] >= 0].copy()

# -----------------------------
# Binary encoding
# -----------------------------
df_pre["iedCode"] = df_pre["iedCode"].map({
    "IMPORT": 0,
    "EXPORT": 1
})

df_pre["reefer"] = df_pre["reefer"].map({
    True: 1,
    False: 0,
    "True": 1,
    "False": 0,
    1: 1,
    0: 0
})

df_pre["released"] = df_pre["released"].map({
    "Y": 1,
    "N": 0,
    1: 1,
    0: 0
})

# -----------------------------
# Frequency encodings
# -----------------------------
arrival_pol_freq = df_raw["arrivalPolCode"].value_counts(normalize=True)
vessel_freq = df_raw["arrivalVesselName"].value_counts(normalize=True)
service_code_freq = df_raw["arrivalServiceCode"].value_counts(normalize=True)

df_pre["arrivalPolCode_freq"] = df_pre["arrivalPolCode"].map(arrival_pol_freq)
df_pre["arrivalVesselName_freq"] = df_pre["arrivalVesselName"].map(vessel_freq)
df_pre["arrivalServiceCode_freq"] = df_pre["arrivalServiceCode"].map(service_code_freq)

# -----------------------------
# ISO top 20 + OTHER
# -----------------------------
top20_iso = df_raw["sizetypeIsoCode"].value_counts().nlargest(20).index

df_pre["sizetypeIsoCode_clean"] = np.where(
    df_pre["sizetypeIsoCode"].isin(top20_iso),
    df_pre["sizetypeIsoCode"],
    "OTHER"
)

df_pre = pd.get_dummies(
    df_pre,
    columns=["sizetypeIsoCode_clean"],
    prefix="iso",
    drop_first=False
)

# -----------------------------
# One-hot encode categorical columns
# -----------------------------
df_pre = pd.get_dummies(
    df_pre,
    columns=[
        "arrivalServiceName",
        "lineCode",
        "arrivalType",
        "typeCode"
    ],
    prefix=[
        "service",
        "line",
        "arrivalType",
        "type"
    ],
    drop_first=False
)

bool_cols = df_pre.select_dtypes(include=["bool"]).columns
df_pre[bool_cols] = df_pre[bool_cols].astype(int)

# -----------------------------
# Drop raw/non-model columns
# -----------------------------
drop_cols = [
    "containerId",
    "customsRefId",
    "mgw",
    "csc",
    "originOfTransportCode",
    "arrivalTime",
    "arrivalPolCode",
    "arrivalVoyageEta",
    "arrivalVesselName",
    "arrivalServiceCode",
    "departureTime",
    "departureType",
    "destinationCode",
    "forwarderName",
    "source_file",
    "location_key",
    "weather_match_time",
    "weather_hour",
    "weather_time",
    "final_departureTime",
    "sizetypeIsoCode"
]

df_pre = df_pre.drop(
    columns=[c for c in drop_cols if c in df_pre.columns]
)

# -----------------------------
# Reorder important columns first
# -----------------------------
front_cols = [
    "remaining_dwell_hours",
    "dwell_hours",
    "elapsed_dwell_hours",
    "iedCode",
    "reefer",
    "teu",
    "gross",
    "arrivalLocationX",
    "arrivalLocationY",
    "arrivalLocationZ",
    "released",
    "numberOfArrivalContainersOnVessel",
    "numberOfRemainOnBoardContainersOnVessel",
    "numberOfDepartureContainersOnVessel",
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "weather_code",
    "cloud_cover",
    "wind_speed_10m",
    "wind_gusts_10m",
    "is_raining",
    "bad_weather_score",
    "arrival_hour",
    "arrival_dayofweek",
    "arrival_month",
    "arrival_is_weekend",
    "arrival_hour_sin",
    "arrival_hour_cos",
    "arrival_month_sin",
    "arrival_month_cos",
    "eta_error_hours",
    "arrivalPolCode_freq",
    "arrivalVesselName_freq",
    "arrivalServiceCode_freq",
    "snapshot_time",
    "arrival_year",
    "snapshot_year",
    "snapshot_date",
    "snapshot_hour",
    "snapshot_minute",
    "snapshot_second"
]

front_cols = [c for c in front_cols if c in df_pre.columns]
other_cols = [c for c in df_pre.columns if c not in front_cols]

df_pre = df_pre[front_cols + other_cols]

# -----------------------------
# Clean missing/infinite values
# -----------------------------
df_pre = df_pre.replace([np.inf, -np.inf], np.nan)

numeric_cols = df_pre.select_dtypes(include=[np.number]).columns
df_pre[numeric_cols] = df_pre[numeric_cols].fillna(df_pre[numeric_cols].median())

# -----------------------------
# Final checks
# -----------------------------
print("\nFinal shape:", df_pre.shape)

print("\nRemaining dwell summary:")
print(df_pre["remaining_dwell_hours"].describe())

print("\nPercent zero remaining dwell:")
print((df_pre["remaining_dwell_hours"] == 0).mean())

print("\nObject columns:")
print(df_pre.select_dtypes(include=["object"]).columns.tolist())

# -----------------------------
# Save
# -----------------------------
df_pre.to_csv(OUTPUT_CSV, index=False)

print("\nSaved:", OUTPUT_CSV)
```
